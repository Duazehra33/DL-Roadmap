```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

np.random.seed(42)
tf.random.set_seed(42)

file_path = "/content/merged_stock_sentiment_data.csv"

df = pd.read_csv(file_path)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
print("\nDataset Columns:")
print(df.columns.tolist())
print("\nFirst Five Rows:")
print(df.head())


df.columns = [
    str(col).strip().lower().replace("_", " ")
    for col in df.columns
]

# Identify Date and Closing Price Columns

date_candidates = ["date", "datetime", "timestamp"]

price_candidates = [
    "adj close",
    "adjusted close",
    "close",
    "closing price"
]

date_column = next(
    (col for col in date_candidates if col in df.columns),
    None
)

price_column = next(
    (col for col in price_candidates if col in df.columns),
    None
)

if date_column is None:
    raise ValueError(
        "Date column not found. Available columns: "
        + str(df.columns.tolist())
    )

if price_column is None:
    raise ValueError(
        "Closing price column not found. Available columns: "
        + str(df.columns.tolist())
    )

ticker_column = next(
    (col for col in ["ticker", "symbol", "stock", "name"]
     if col in df.columns),
    None
)

if ticker_column is not None:

    print("\nAvailable Stocks:")
    print(df[ticker_column].value_counts().head(20))

    selected_stock = None


    if selected_stock is None:
        selected_stock = df[ticker_column].value_counts().index[0]

    df = df[
        df[ticker_column].astype(str).str.upper()
        == str(selected_stock).upper()
    ].copy()

    print("\nSelected Stock:", selected_stock)

# Data Preprocessing

df[date_column] = pd.to_datetime(
    df[date_column],
    errors="coerce"
)

df[price_column] = pd.to_numeric(
    df[price_column].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

df = df[[date_column, price_column]].copy()

df.columns = ["Date", "Close"]

df = df.dropna()

df = df.sort_values("Date")

df = df.drop_duplicates(subset="Date")

df = df.set_index("Date")

if len(df) < 150:
    raise ValueError(
        "Insufficient stock records. At least 150 records are required."
    )

print("\nCleaned Dataset:")
print(df.head())
print("\nFinal Dataset Shape:", df.shape)

#  Visualize Historical Stock Prices

plt.figure(figsize=(14, 5))

plt.plot(df.index, df["Close"])

plt.title("Historical Stock Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.grid()
plt.show()

#  Normalize Dataset

train_size = int(len(df) * 0.8)

scaler = MinMaxScaler(feature_range=(0, 1))

scaler.fit(df[["Close"]].iloc[:train_size])

scaled_data = scaler.transform(df[["Close"]])

train_data = scaled_data[:train_size]

test_data = scaled_data[train_size - 60:]

print("\nTraining Data Shape:", train_data.shape)
print("Testing Data Shape:", test_data.shape)

# Create Time-Series Sequences

time_steps = 60

def create_sequences(data, time_steps=60):

    X = []
    y = []

    for i in range(time_steps, len(data)):

        X.append(data[i-time_steps:i, 0])
        y.append(data[i, 0])

    return np.array(X), np.array(y)


X_train, y_train = create_sequences(
    train_data,
    time_steps
)

X_test, y_test = create_sequences(
    test_data,
    time_steps
)

X_train = X_train.reshape(
    X_train.shape[0],
    X_train.shape[1],
    1
)

X_test = X_test.reshape(
    X_test.shape[0],
    X_test.shape[1],
    1
)

print("\nSequence Shapes:")

print("X Train:", X_train.shape)
print("y Train:", y_train.shape)

print("X Test:", X_test.shape)
print("y Test:", y_test.shape)

# Build LSTM Model

lstm_model = Sequential([

    Input(shape=(time_steps, 1)),

    LSTM(50, return_sequences=True),

    Dropout(0.2),

    LSTM(50),

    Dropout(0.2),

    Dense(25, activation="relu"),

    Dense(1)

])

lstm_model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

print("\nLSTM Model Summary")

lstm_model.summary()

# Train LSTM Model

lstm_early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

lstm_history = lstm_model.fit(

    X_train,
    y_train,

    epochs=30,

    batch_size=32,

    validation_split=0.1,

    shuffle=False,

    callbacks=[lstm_early_stop]

)

# Build GRU Model

gru_model = Sequential([

    Input(shape=(time_steps, 1)),

    GRU(50, return_sequences=True),

    Dropout(0.2),

    GRU(50),

    Dropout(0.2),

    Dense(25, activation="relu"),

    Dense(1)

])

gru_model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

print("\nGRU Model Summary")

gru_model.summary()

#Train GRU Model

gru_early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

gru_history = gru_model.fit(

    X_train,
    y_train,

    epochs=30,

    batch_size=32,

    validation_split=0.1,

    shuffle=False,

    callbacks=[gru_early_stop]

)

# Plot Training and Validation Loss

plt.figure(figsize=(14, 5))

plt.plot(
    lstm_history.history["loss"],
    label="LSTM Training Loss"
)

plt.plot(
    lstm_history.history["val_loss"],
    label="LSTM Validation Loss"
)

plt.plot(
    gru_history.history["loss"],
    label="GRU Training Loss"
)

plt.plot(
    gru_history.history["val_loss"],
    label="GRU Validation Loss"
)

plt.title("LSTM vs GRU Training Performance")

plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.legend()
plt.grid()
plt.show()

# Make LSTM Predictions

lstm_predictions = lstm_model.predict(X_test)

lstm_predictions = scaler.inverse_transform(
    lstm_predictions
)

# Make GRU Predictions

gru_predictions = gru_model.predict(X_test)

gru_predictions = scaler.inverse_transform(
    gru_predictions
)

# Convert Actual Prices to Original Scale

actual_prices = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)

test_dates = df.index[train_size:]

print("\nFirst Five Actual Prices:")
print(actual_prices[:5])

print("\nFirst Five LSTM Predictions:")
print(lstm_predictions[:5])

print("\nFirst Five GRU Predictions:")
print(gru_predictions[:5])

# Plot Actual vs LSTM Predictions

plt.figure(figsize=(14, 6))

plt.plot(
    test_dates,
    actual_prices.flatten(),
    label="Actual Prices"
)

plt.plot(
    test_dates,
    lstm_predictions.flatten(),
    label="LSTM Predictions"
)

plt.title("Stock Price Prediction Using LSTM")

plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Plot Actual vs GRU Predictions

plt.figure(figsize=(14, 6))

plt.plot(
    test_dates,
    actual_prices.flatten(),
    label="Actual Prices"
)

plt.plot(
    test_dates,
    gru_predictions.flatten(),
    label="GRU Predictions"
)

plt.title("Stock Price Prediction Using GRU")

plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Compare LSTM and GRU Predictions

plt.figure(figsize=(14, 6))

plt.plot(
    test_dates,
    actual_prices.flatten(),
    label="Actual Prices"
)

plt.plot(
    test_dates,
    lstm_predictions.flatten(),
    label="LSTM Predictions"
)

plt.plot(
    test_dates,
    gru_predictions.flatten(),
    label="GRU Predictions"
)

plt.title("Stock Price Prediction: LSTM vs GRU")

plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Evaluate LSTM Model

lstm_mse = mean_squared_error(
    actual_prices,
    lstm_predictions
)

lstm_mae = mean_absolute_error(
    actual_prices,
    lstm_predictions
)

lstm_rmse = np.sqrt(lstm_mse)

# Evaluate GRU Model

gru_mse = mean_squared_error(
    actual_prices,
    gru_predictions
)

gru_mae = mean_absolute_error(
    actual_prices,
    gru_predictions
)

gru_rmse = np.sqrt(gru_mse)

#  Compare Model Performance

results = pd.DataFrame({

    "Model": ["LSTM", "GRU"],

    "MSE": [lstm_mse, gru_mse],

    "MAE": [lstm_mae, gru_mae],

    "RMSE": [lstm_rmse, gru_rmse]

})

print("\nMODEL PERFORMANCE COMPARISON")

print(results.to_string(index=False))

# Identify Better Model Based on RMSE

if lstm_rmse < gru_rmse:
    print("\nLSTM performed better based on RMSE.")
else:
    print("\nGRU performed better based on RMSE.")

prediction_results = pd.DataFrame({

    "Date": test_dates,

    "Actual Price": actual_prices.flatten(),

    "LSTM Prediction": lstm_predictions.flatten(),

    "GRU Prediction": gru_predictions.flatten()

})

prediction_results.to_csv(
    "/content/stock_price_predictions.csv",
    index=False
)

print("\nPrediction Results Saved Successfully!")

print(prediction_results.head())

```
